```python
from torchwnn.datasets.iris import Iris
from torchwnn.classifiers import Wisard
from torchwnn.encoding import Thermometer
import pandas as pd

from os import listdir
from os.path import isfile  , join
```

```python
pasta_csv="../../../dataset"
n_classes=[]
for arquivo in sorted(listdir(pasta_csv),key=lambda s: (len(s), s)):
    file_path=join(pasta_csv, arquivo)
    if isfile(file_path) and ".csv" in arquivo:
        df=pd.read_csv(file_path)
        
        #retirando IPV6 e preparando IPV4 para int
        df = df[~df["src_ip"].str.contains(":", na=False)]
        df = df[~df["dst_ip"].str.contains(":", na=False)]
        df['src_ip']=df['src_ip'].str.replace('.','')
        df['dst_ip']=df['dst_ip'].str.replace('.','')
        
        n_classes= list(set(n_classes) | set(df['label'].unique().tolist()))
        features=df[["ts","src_ip","src_port","dst_ip","dst_port","duration","src_bytes","dst_bytes","missed_bytes","src_pkts","src_ip_bytes","dst_pkts","dst_ip_bytes"]].astype("float")
        labels=df[["label","type"]]
        
        features.to_pickle("./data/"+arquivo.replace(".csv",".pkl").replace("Network_dataset","feature"))
        labels.to_pickle("./data/"+arquivo.replace(".csv",".pkl").replace("Network_dataset","label"))
pd.DataFrame([len(n_classes)],columns=['n_classes']).to_pickle('./data/n_classes.pkl')
#pd.read_pickle("dados.pkl")
```
