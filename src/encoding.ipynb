```python
import math

def log_thermometer_encoding(value, size=8, base=7):
    """
    Encode um valor usando uma técnica de log-thermometer.
    log(x+1) then to binary
    """
    if value <= 0:
        raise ValueError("O valor deve ser positivo para logaritmo.")
    
    log_value = math.log(value+1, base)
    
    level = int(math.ceil(log_value))
    
    
    # sequência de bits '1' seguida de zeros, com comprimento igual a 'level'
    encoded_bits = '1' * level
    
    # Preenche com zeros à esquerda para atingir o tamanho fixo
    encoded_bits = encoded_bits.ljust(size, '0')
    return f'{encoded_bits}'[:size]


valor = 4444451
codigo_binario = log_thermometer_encoding(valor)
print(f"Valor: {valor}")
print(f"Código binário (Log-Thermometer): {codigo_binario}")
```

```python
def p_encoding(p):
    """
    Encode udp/tcp unicamente
    """
    dict={"TCP":"0011","UDP":"1100"}
    if p in dict.keys():
        return dict[p]
    else:
        return "0000"

p="TCP"
print(f"p: {p}")
print(f"Binary : {p_encoding(p)}")
```

```python
def ip_encoding(ip):
    """
    Encode 2 primeiros octetos do ip unicamente
    """
    ip= ip.split('.')
    return bin(int(ip[0]))[2:].zfill(8) + bin(int(ip[1]))[2:].zfill(8)

ip ='192.168.0.0'
print(f"ip: {ip}")
print(f"Binary : {ip_encoding(ip)}")
```
